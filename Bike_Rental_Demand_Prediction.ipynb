````python
# ============================================
# Day 17: Data Cleaning & Preprocessing
# Project: Bike Rental Demand Prediction
# Dataset: UCI Bike Sharing Dataset
# ============================================

# Import required libraries
import pandas as pd
import zipfile
import os
````

# ***`Block 7 — Remove Duplicate Records`***

````python
# --------------------------------------------
# Step 6: Remove Duplicate Records
# --------------------------------------------

before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
````

# ***`Block 8 — Check Data Types`***

````python
# --------------------------------------------
# Step 7: Check Data Types
# --------------------------------------------

print("=== Data Types Before Preprocessing ===")

print(df.dtypes)
````

# ***`TIME FEATURE PROCESSING`***
# ***Block 9 — Convert Date Column***

````python
# --------------------------------------------
# Step 8: Convert dteday to datetime
# --------------------------------------------
# dteday contains the date information.
# Convert it from text/object into datetime.

df["dteday"] = pd.to_datetime(df["dteday"])

print("Date column converted successfully!")

print(df["dteday"].head())
````

# ***`Block 10 — Create Time-Based Features`***

````python
# --------------------------------------------
# Step 9: Create time-based features
# --------------------------------------------
# Extract useful information from the date.

df["day"] = df["dteday"].dt.day
df["month"] = df["dteday"].dt.month
df["date_year"] = df["dteday"].dt.year
````

# ***`Block 11 — Create Cyclical Hour Features`***

````python
# --------------------------------------------
# Step 10: Process hour as cyclical data
# --------------------------------------------
# Hour is cyclical:
# 23 and 0 are close to each other.
# Sin/Cos encoding represents this relationship.

import numpy as np

df["hr_sin"] = np.sin(2 * np.pi * df["hr"] / 24)
df["hr_cos"] = np.cos(2 * np.pi * df["hr"] / 24)

print("Time features created successfully!")
````

# ***`CATEGORICAL ENCODING`***
# ***Block 12 — Encode Categorical Features***
# `*The UCI dataset uses numeric codes for several categorical variables. We can convert them into one-hot encoded columns.*`

````python
# --------------------------------------------
# Step 11: One-Hot Encoding
# --------------------------------------------
# Convert categorical features into numerical
# dummy variables.

categorical_columns = [
    "season",
    "mnth",
    "hr",
    "weekday",
    "weathersit"
]

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

print("Categorical features encoded successfully!")
````

# ***`Block 13 — Convert Boolean Columns to Integers`***

````python
# --------------------------------------------
# Step 12: Convert boolean columns to 0/1
# --------------------------------------------
# One-hot encoding may create True/False columns.
# Convert them into numerical 1/0 values.

bool_columns = df.select_dtypes(include="bool").columns

df[bool_columns] = df[bool_columns].astype(int)

print("Boolean values converted to numerical values.")
````

# ***`REMOVE UNNECESSARY / LEAKAGE COLUMNS`***
# ***Block 14 — Remove Unnecessary Columns***

````python
# --------------------------------------------
# Step 13: Remove unnecessary columns
# --------------------------------------------
# instant       -> only an ID/index
# dteday        -> original date after extracting features
# casual        -> component of target
# registered    -> component of target
#
# casual + registered = cnt
# Keeping them would give the model information
# directly related to the target.

columns_to_remove = [
    "instant",
    "dteday",
    "casual",
    "registered"
]

df = df.drop(columns=columns_to_remove)

print("Unnecessary and target-leakage columns removed.")
````

# ***`Block 1 — Import Evaluation Metrics`***

````python
# ============================================
# Day 21: Model Evaluation
# Project: Bike Rental Demand Prediction
# ============================================

# Import regression evaluation metrics

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import matplotlib.pyplot as plt
````

# ***`Block 2 — Calculate MAE`***

````python
# --------------------------------------------
# Step 1: Mean Absolute Error (MAE)
# --------------------------------------------
# MAE shows the average absolute difference
# between actual and predicted rental counts.

mae = mean_absolute_error(y_test, y_pred)

print("=== Mean Absolute Error (MAE) ===")
print(mae)
````

# ***`Block 3 — Create the Input Preparation Function`***

````python
# --------------------------------------------
# Step 2: Prepare new input for the model
# --------------------------------------------

def prepare_input(
    year,
    season,
    month,
    hour,
    holiday,
    weekday,
    workingday,
    weather,
    temp,
    atemp,
    hum,
    windspeed,
    day
):
    # Convert selected year into the same format
    # used by the original dataset.
    yr = 0 if year == "2011" else 1
    date_year = int(year)

    # Create one new data record
    new_data = pd.DataFrame([{
        "yr": yr,
        "holiday": holiday,
        "workingday": workingday,
        "temp": temp,
        "atemp": atemp,
        "hum": hum,
        "windspeed": windspeed,
        "season": season,
        "mnth": month,
        "hr": hour,
        "weekday": weekday,
        "weathersit": weather,
        "day": day,
        "month": month,
        "date_year": date_year
    }])

    # Create cyclical hour features
    new_data["hr_sin"] = np.sin(2 * np.pi * new_data["hr"] / 24)
    new_data["hr_cos"] = np.cos(2 * np.pi * new_data["hr"] / 24)

    # One-hot encode categorical features
    categorical_columns = [
        "season",
        "mnth",
        "hr",
        "weekday",
        "weathersit"
    ]

    new_data = pd.get_dummies(
        new_data,
        columns=categorical_columns,
        drop_first=True
    )

    # Convert Boolean columns into 0/1
    bool_columns = new_data.select_dtypes(
        include="bool"
    ).columns

    new_data[bool_columns] = new_data[bool_columns].astype(int)

    # Make the columns exactly the same as
    # the columns used when training the model.
    new_data = new_data.reindex(
        columns=X.columns,
        fill_value=0
    )

    return new_data
````

````python
# --------------------------------------------
# Step 3: Create Prediction Function
# --------------------------------------------

def predict_demand(
    year,
    season,
    month,
    hour,
    holiday,
    weekday,
    workingday,
    weather,
    temp,
    atemp,
    hum,
    windspeed,
    day
):
    # Prepare the new input
    input_data = prepare_input(
        year,
        season,
        month,
        hour,
        holiday,
        weekday,
        workingday,
        weather,
        temp,
        atemp,
        hum,
        windspeed,
        day
    )

    # Generate prediction
    prediction = model.predict(input_data)[0]

    # Keep the displayed demand non-negative.
    # This is only for the application output;
    # the original model prediction remains unchanged
    # for Day 21 evaluation.
    prediction = max(0, prediction)

    return round(prediction, 2)
````
